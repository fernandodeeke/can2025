Criemos uma função que soma linhas:

```python
def sum_rows(A,k, i, j):
  # Add k time row j to row i
  n = A.shape[0]
  E = np.eye(n)
  if i == j:
    E[i,i] = k+1
  else:
    E[i,j] = k
  return E@A
```

Suponha que queremos somente multiplicar uma linha por um escalar:

```python
def scale_row(A,k,i):
  # Multiply row i by k
  n = A.shape[0]
  E = np.eye(n)
  E[i,i] = k
  return E@A
```

Troca de linhas:

```python
def swap_rows(A,i,j):
  # Swap rows i and j
  n = A.shape[0]
  E = np.eye(n)
  E[i,i] = 0
  E[j,j] = 0
  E[i,j] = 1
  E[j,i] = 1
  return E@A
```
