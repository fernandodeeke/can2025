```python
import numpy as np
from  scipy.interpolate import CubicSpline
```

```python
t = np.array([0,1,2,3,4,5])
y = np.array([0,3,5,7,4,1])
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,4))
plt.scatter(t,y, color = 'black')
plt.show()
```

```python
cs_clamped = CubicSpline(t, y, bc_type=((1,2.5), (1, -1.0)))
cs_natural = CubicSpline(t, y, bc_type='natural')
cs_not_a_knot = CubicSpline(t, y, bc_type='not-a-knot')
```

```python
t_suave = np.linspace(-0.5, 5.5, 100)
```

```python
y_clamped = cs_clamped(t_suave)
y_natural = cs_natural(t_suave)
y_not_a_knot = cs_not_a_knot(t_suave)
```

```python
plt.figure(figsize=(8,4))
plt.scatter(t,y, color = 'black')
plt.plot(t_suave, y_clamped, label='clamped')
plt.plot(t_suave, y_natural, label='natural')
plt.plot(t_suave, y_not_a_knot, label='not-a-knot')
plt.legend()
plt.show()
```

```python
coef_clamped = cs_clamped.c
coef_natural = cs_natural.c
coef_not_a_knot = cs_not_a_knot.c
```
