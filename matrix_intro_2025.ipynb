Consider the vectors $\mathbf{u}=(3,4,2)$ and $\mathbf{v}=(-1,2,-1)$. The scalar product $\mathbf{u}\cdot \mathbf{v}$ is given by

```python
u = np.array([3,4,2])
v = np.array([-1,2,-1])
np.dot(u,v)
```

The vector product $\mathbf{u} \times \mathbf{v}$ is given by

```python
np.cross(u,v)
```

The Frobenius (or Euclidean) norm of a vector $\mathbf{u}=(u_1,\ldots,u_n)$ is defined by
$$
|\mathbf{u}|=\left[\sum\limits_{i=1}^{n}|u_i|^2\right]^{1/2}
$$
In Python:

```python
u = np.array([4,-6,1])
u
```

The Euclidean norm of u is given by:

```python
LA.norm(u)
```

The so-called "infinite norm" of a vector is defined as the largest magnitude among the components:

$$
|\mathbf{u}|_{\infty}=\max_i\{{|u_i|},i=1,\ldots,n\}
$$

In Python,

```python
LA.norm(u, np.inf)
```

The infinite standard is the most widely used in numerical computing.

**Example 1**. Let us calculate the projection of the vector $\mathbf{u}$ in the direction of $\mathbf{v}$.

We must calculate the expression

$$
\mathbf{u}_{\mathbf{v}}=\left(\frac{\mathbf{u}\cdot\mathbf{v}}{|\mathbf{v}|^2}\right)\mathbf{v}
$$

In Python,

```python
uv = (np.dot(u,v)/np.dot(v,v))*v
uv
```

The Frobenius norm of a matrix $A=[a_{ij}]$ of order $m \times n$ is defined by

$$
|A|_F=
\left[
\sum\limits_{i=1}^{m}
\sum\limits_{j=1}^{n}
|a_{ij}|^2
\right]^{1/2}
$$

In Python,

```python
A = np.array([[1,  1,  2],[ 0,  -1,  -1],[ 2,  1,  0]])
A
```

```python
LA.norm(A,'fro')
```

The infinite norm of an array $A=[a_{ij}]$ of order $m \times n$ is often used in iterative process stop criteria is defined as the maximum value of the absolute sum of each row

$$
|A|_{\infty}= \max_{1\leq i\leq m}\sum_{j}^n|a_{ij}|\,.
$$

In Python,

```python
LA.norm(A, np.inf)
```

Note that when the type of norm is not specified the default result is the Frobenius norm:

```python
LA.norm(A)
```

**1**. Calculate the area of a triangle of vertices $(-1,3,2)$, $(3,4,-1)$, and $(3,5,1)$. Suggestion: Use the geometric interpretation of the vector product.

**Solution**. By defining the position vectors $a = (-1,2,3)$, $b= (3,4,-1)$ and $c=(3,5,1)$, and defining the vectors $u=b-a$ and $v=c-a$, the area of the triangle defined by these points can be calculated by

$$
A = \frac{1}{2}\left|u \times v \right|,.
$$

Let's do this calculation in Python.

```python
import numpy as np
import numpy.linalg as LA
```

```python
a = np.array([-1,3,3])
b = np.array([3,4,-1])
c = np.array([3,5,1])
u = b-a
v = c-a
```

```python
A = 0.5*LA.norm(np.cross(u,v))
A
```

Therefore, the area of the triangle is approximately equal to 5.39 units of area.

**2**. Calculate the volume of the parallelepiped determined by the vectors $u=(1,5,4)$, $v=(2,4,1)$ and $w=(9,2,6)$.

**Solution.** The volume is given by

$$
V = |u \cdot (v \times w)|,.
$$

In python,

```python
import numpy as np
import numpy.linalg as LA
```

```python
u = np.array([1,5,4])
v = np.array([2,4,1])
w = np.array([9,2,6])
```

```python
V1 = np.cross(u,v)
```

```python
V2 = np.abs(np.dot(w,V1))
V2
```

Ou seja, 121 unidades de volume.
